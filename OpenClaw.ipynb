```python
import transformers
import accelerate
import torch

# Install dependencies
!pip -q install transformers accelerate torch

from transformers import pipeline
import random

print("Loading AI Model...")

# Use supported task
llm = pipeline(
    "text-generation",
    model="google/flan-t5-base",
    max_new_tokens=100
)

print("Model loaded successfully")

# --------------------------------------------------
# Simulated Infrastructure
# --------------------------------------------------

servers = {
    "web-server": {"cpu": 85, "memory": 70},
    "db-server": {"cpu": 40, "memory": 60},
    "app-server": {"cpu": 92, "memory": 80}
}

logs = [
    "ERROR: Database connection timeout",
    "WARNING: CPU usage exceeded 90%",
    "ERROR: Disk space critical on /var",
    "INFO: Service restarted successfully"
]

# --------------------------------------------------
# IT Operations Tools
# --------------------------------------------------

def health_check():
    server = random.choice(list(servers.keys()))
    cpu = servers[server]["cpu"]
    mem = servers[server]["memory"]
    return f"{server} -> CPU {cpu}% | Memory {mem}%"

def log_analysis():
    return random.choice(logs)

def recommend_fix(issue):

    prompt = f"Suggest troubleshooting steps for this IT issue: {issue}"

    response = llm(prompt)[0]["generated_text"]

    return response

def summarize_incident(text):

    prompt = f"Summarize this IT incident briefly: {text}"

    response = llm(prompt)[0]["generated_text"]

    return response

# --------------------------------------------------
# Tool Registry
# --------------------------------------------------

tools = {
    "health_check": health_check,
    "log_analysis": log_analysis,
    "recommend_fix": recommend_fix,
    "incident_summary": summarize_incident
}

# --------------------------------------------------
# Agent Router
# --------------------------------------------------

def choose_tool(user_input):

    prompt = f"""
Choose the correct IT operations tool.

Tools:
health_check
log_analysis
recommend_fix
incident_summary

User request: {user_input}

Answer with tool name only.
"""

    response = llm(prompt)[0]["generated_text"].lower()

    for tool in tools:
        if tool in response:
            return tool

    return None

# --------------------------------------------------
# Agent Loop
# --------------------------------------------------

print("\nAI ITOps Agent Started")
print("Type 'exit' to stop\n")

while True:

    user_input = input("User: ")

    if user_input.lower() == "exit":
        print("Agent stopped.")
        break

    tool = choose_tool(user_input)

    print("Selected tool:", tool)

    if tool == "health_check":
        result = health_check()

    elif tool == "log_analysis":
        result = log_analysis()

    elif tool == "recommend_fix":
        result = recommend_fix(user_input)

    elif tool == "incident_summary":
        result = summarize_incident(user_input)

    else:
        result = "No suitable tool identified."

    print("Agent:", result)
```

```python
# =========================
# OPENCLAW-STYLE AGENT (COLAB - SINGLE CELL)
# =========================

# Install dependencies
!pip -q install transformers accelerate bitsandbytes

# Imports
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# Load model (lightweight + Colab friendly)
model_name = "mistralai/Mistral-7B-Instruct-v0.1"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    load_in_4bit=True
)

# -------------------------
# Simple Tool (simulate search)
# -------------------------
def search_tool(query):
    return f"[Search Results]: Top insights about '{query}' → Identity risk, admin misuse, lateral movement."

# -------------------------
# OpenClaw-style Agent
# -------------------------
def openclaw_agent(task):
    prompt = f"""
You are an autonomous AI agent.

Follow this structure:
1. PLAN: Break the problem
2. ACTION: Decide what to do
3. OBSERVE: Use tool if needed
4. FINAL ANSWER: Clear output

Task: {task}
"""

    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

    output = model.generate(
        **inputs,
        max_new_tokens=300,
        temperature=0.7
    )

    response = tokenizer.decode(output[0], skip_special_tokens=True)

    # Simulated tool usage trigger
    if "search" in response.lower():
        tool_output = search_tool(task)
        response += f"\n\nTOOL OUTPUT:\n{tool_output}"

    return response

# -------------------------
# Run Example
# -------------------------
task = "What are top risks of giving full admin access to engineering teams?"

result = openclaw_agent(task)

print(result)
```

```python
# =========================
# OPENCLAW-STYLE AGENT (FIXED - COLAB SINGLE CELL)
# =========================

# Install dependencies
!pip -q install transformers accelerate bitsandbytes

# Imports
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
import torch

# -------------------------
# Quantization Config (FIX)
# -------------------------
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4"
)

# -------------------------
# Load Model
# -------------------------
model_name = "mistralai/Mistral-7B-Instruct-v0.1"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    quantization_config=bnb_config
)

# -------------------------
# Simple Tool
# -------------------------
def search_tool(query):
    return f"[Search Results]: Risks include privilege escalation, malware install, lateral movement."

# -------------------------
# OpenClaw-style Agent
# -------------------------
def openclaw_agent(task):
    prompt = f"""
You are an autonomous AI agent.

Follow:
PLAN → ACTION → OBSERVE → FINAL ANSWER

Task: {task}
"""

    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

    output = model.generate(
        **inputs,
        max_new_tokens=300,
        temperature=0.7
    )

    response = tokenizer.decode(output[0], skip_special_tokens=True)

    if "search" in response.lower():
        response += "\n\nTOOL OUTPUT:\n" + search_tool(task)

    return response

# -------------------------
# Run Example
# -------------------------
print(openclaw_agent("What are risks of giving admin access to engineers?"))
```
